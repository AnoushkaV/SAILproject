```python
import os
import pandas as pd

# open file
file = os.path.join('..', 'Data', 'sensor-location.xlsx')
df = pd.read_excel(file)

# 1. Kolomnamen opschonen: onzichtbare \xa0 vervangen door een gewone spatie
df.columns = df.columns.str.replace('\xa0', ' ', regex=False).str.strip()
print(df.columns.tolist())   # check of 'Effectieve breedte' nu normaal is

id_col = df.columns[0]       # kolom met de sensor-ID's (pas aan indien nodig)
df[id_col] = df[id_col].astype(str).str.strip()

# 2. GASA-06-B koppelen aan de locatie van GASA-06 (Azartplein B)
gasa06 = df[df[id_col].str.endswith('GASA-06')].copy()
gasa06[id_col] = gasa06[id_col].str.replace('GASA-06', 'GASA-06-B', regex=False)
df = pd.concat([df, gasa06], ignore_index=True)

# 3. Defecte sensoren GASA-06 en GVCV-09 verwijderen (zitten niet meer in de tellingen)
defect = df[id_col].str.endswith('GASA-06') | df[id_col].str.endswith('GVCV-09')
df = df[~defect].reset_index(drop=True)

# controle
print(df[df[id_col].str.contains('GASA|GVCV')])

# look for duplicates
duplicates = df[df.duplicated()]
print(duplicates)

# look for empty values
df[df.isna().any(axis=1)]
```
