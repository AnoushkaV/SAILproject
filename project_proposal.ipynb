# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

To develop a dashboard that helps crowd managers identify peak visitor flows and sudden changes in pedestrian flows at selected sensor locations during SAIL2025.

* At which selected locations and times were the highest visitor flows recorded during SAIL2025?

* How did visitor flows at these locations change throughout the event?

* Where and when did sudden increases or decreases in pedestrian flows occur?

* How can a dashboard help crowd managers spot peaks and sudden changes quickly and correctly?

# Data Used

The main datasets used in this study are: 
SAIL2025_LVMA_3min_20August-25August2025_flow
sensor-location

We will use pedestrian counts from the LVMA sensor dataset, recorded at three-minute intervals during SAIL2025. These measurements will be combined with the sensor-location.xlsx dataset, which provides the geographical locations of the sensors and their effective pedestrian widths.

This combination allows us to examine absolute pedestrian flows as well as pedestrian flow per metre of effective walking width. We will use these indicators to compare crowd conditions between selected locations, identify peak periods and detect sudden changes over time.

Where a sensor records pedestrian movements in two directions, we will first examine the directional measurements separately. Depending on the research question and sensor configuration, these flows may subsequently be combined to represent the total pedestrian flow at that location.


# Data Pipeline

First, we will load the LVMA pedestrian counts and sensor location data in Python. We will conduct a preliminary analysis to inspect the available sensors, measurement period and data quality. Next, we will clean the measurements, select the SAIL2025 event period and link the counts to their sensor locations and effective widths. We will then calculate and compare visitor flows, identify peaks and sudden changes over time, and present the results in dashboard visualisations. 


First, the LVMA pedestrian count data and sensor-location data will be imported into Python. An exploratory analysis will be performed to understand the available sensors, variables, measurement period, measurement frequency and overall data quality.

The following analysis pipeline will then be used:
1. Data inspection and cleaning
Check for missing values, duplicated measurements, incorrect timestamps, inconsistent sensor identifiers and potential outliers.

2. Temporal selection
Filter the measurements to the SAIL2025 event period from 20 to 24 August 2025.

3. Data integration
Link pedestrian measurements to the corresponding sensor locations and effective pedestrian widths.

4. Indicator calculation
Calculate indicators such as:
- pedestrian count per three-minute interval;
- pedestrian flow per metre of effective walking width;
- peak pedestrian flow per sensor;
- time of peak flow;
- changes in pedestrian flow between consecutive measurements.

5. Peak and change detection
Identify the locations and time periods with the highest visitor flows. Sudden increases and decreases will be analysed by examining changes between consecutive time intervals and deviations from typical flow patterns at each sensor.

6. Comparison across locations and time
Compare flow patterns between selected sensor locations, event days and different periods of the day.

7. Dashboard visualisation
Present the results using clear visualisations such as time-series graphs, peak-flow indicators, maps and location comparisons. The dashboard will be designed to allow crowd managers to quickly identify locations and periods associated with high or rapidly changing visitor flows.




**Geographical Scale**
The geographical scope of the study is the SAIL2025 event area in Amsterdam. The analysis will focus on selected LVMA pedestrian sensor locations within and around the main SAIL event area.

**Temporal Scale**
The analysis covers the main SAIL2025 event period from 20 August to 24 August 2025. The available dataset contains measurements from 20 to 25 August 2025, but measurements outside the main event period will only be used if they are relevant as a reference or for data validation.

**Intended Output**
The main output of the project will be an interactive crowd-monitoring dashboard. The dashboard will show how pedestrian flows varied across selected locations and over time during SAIL2025. It should enable crowd managers to quickly identify peak crowd periods, locations experiencing relatively high pedestrian pressure and sudden changes in visitor flows.

The analysis will therefore provide both a retrospective assessment of crowd patterns during SAIL2025 and a dashboard structure that could potentially support crowd monitoring during similar large-scale events.
